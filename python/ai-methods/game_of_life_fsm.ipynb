```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from enum import Enum
from dataclasses import dataclass
import time

class CellState(Enum):
    DEAD = 0 
    ALIVE = 1

@dataclass
class Cell:
    x: int
    y: int 
    state: CellState = CellState.DEAD
    
    def is_alive(self):
        return self.state == CellState.ALIVE
        
    def set_state(self, new_state: CellState):
        self.state = new_state

class GameOfLife:
    def __init__(self, width: int, height: int):
        self.width = width
        self.height = height
        self.grid = [[Cell(x, y) for x in range(width)] for y in range(height)]
        
    def get_neighbors(self, x: int, y: int):
        neighbors = []
        for dx in [-1, 0, 1]:
            for dy in [-1, 0, 1]:
                if dx == 0 and dy == 0:
                    continue
                new_x = (x + dx) % self.width
                new_y = (y + dy) % self.height
                neighbors.append(self.grid[new_y][new_x])
        return neighbors
        
    def next_generation(self):
        new_states = {}
        for y in range(self.height):
            for x in range(self.width):
                cell = self.grid[y][x]
                alive_neighbors = sum(1 for n in self.get_neighbors(x, y) if n.is_alive())
                
                if cell.is_alive():
                    new_states[(x,y)] = CellState.ALIVE if alive_neighbors in [2, 3] else CellState.DEAD
                else:
                    new_states[(x,y)] = CellState.ALIVE if alive_neighbors == 3 else CellState.DEAD
                    
        for (x, y), state in new_states.items():
            self.grid[y][x].set_state(state)
            
    def randomize(self):
        for row in self.grid:
            for cell in row:
                cell.set_state(CellState.ALIVE if np.random.random() < 0.3 else CellState.DEAD)
                
    def get_grid_state(self):
        return np.array([[1 if cell.is_alive() else 0 for cell in row] for row in self.grid])
    
def animate(self):
    grid.next_generation()
    state_grid = grid.get_grid_state()
    img = ax.imshow(state_grid, cmap='binary', interpolation='nearest')
    img.set_data(grid.get_grid_state())
    return img,
    

grid = GameOfLife(250, 250)
rand = grid.randomize()
fig, ax = plt.subplots()
img = ax.imshow(grid.get_grid_state(), cmap='binary', interpolation='nearest')
ax.set_title("Игра жизнь")
ani = animation.FuncAnimation(fig, animate, frames=100, interval=200, blit=True)    
plt.show()
```
