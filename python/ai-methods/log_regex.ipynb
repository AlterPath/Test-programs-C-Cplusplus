```python
import re
```

```python
f = open('Reg.txt','r', encoding='utf-8')
f = f.readlines()
f = list(repr(f).split('\n'))
```

```python
for line in f:
    data_a = re.search(r'\d{4}-\d\d-\d\d', line)
    data_b = re.search(r'\d\d:\d\d:\d\d', line)
    tpe = re.search(r'\[(\w+)\]', line)
    ip = re.search(r'\d+.\d+\.\d+\.\d+', line)
    print(data_a, data_b,tpe, ip)
```
